```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\query prossesing\alphabet_stock_data.csv")

print(df)

df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

data = df[(df["Date"] >= "2020-04-01") & (df["Date"] <= "2020-04-30")]

plt.figure(figsize=(10,5))
plt.plot(data["Date"], data["Close"], marker='o')
plt.title("Historical Stock Prices of Alphabet Inc.")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv(r"C:\query prossesing\alphabet_stock_data.csv")

# Display the data
print(df)

# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Select records between two dates
data = df[(df["Date"] >= "2020-04-01") &
          (df["Date"] <= "2020-04-30")]

# Create Scatter Plot
plt.figure(figsize=(10,5))
plt.scatter(data["Volume"], data["Close"], color="blue")

# Title and labels
plt.title("Scatter Plot of Trading Volume vs Stock Price")
plt.xlabel("Trading Volume")
plt.ylabel("Closing Stock Price")

# Display Grid
plt.grid(True)

# Show Plot
plt.show()
```

```python
import numpy as np

df = pd.DataFrame(
    np.random.randn(10, 4),
    columns=['A', 'B', 'C', 'D']
)

def color_values(val):
    if val < 0:
        return 'color: red'
    else:
        return 'color: black'

styled_df = df.style.map(color_values)

print(df)

styled_df
```

```python
df = pd.DataFrame(
    np.random.randn(10, 4),
    columns=['A', 'B', 'C', 'D']
)

df.iloc[1, 2] = np.nan
df.iloc[4, 0] = np.nan
df.iloc[7, 3] = np.nan

def highlight_nan(val):
    if pd.isna(val):
        return 'background-color: yellow'
    return ''

styled_df = df.style.map(highlight_nan)

print(df)

styled_df
```

```python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    'school_code': ['s001', 's002', 's001', 's003', 's002'],
    'student': ['A', 'B', 'C', 'D', 'E'],
    'age': [15, 16, 15, 17, 16]
})

# Split into groups based on school_code
group = df.groupby('school_code')

# Display type
print("Type of GroupBy object:")
print(type(group))

# Display groups
print("\nGroups:")
for name, data in group:
    print("\nSchool Code:", name)
    print(data)
```

```python
df = pd.DataFrame({
    'school_code': ['s001', 's002', 's001', 's003', 's002'],
    'student': ['A', 'B', 'C', 'D', 'E'],
    'age': [15, 16, 17, 18, 15]
})

result = df.groupby('school_code')['age'].agg(['mean', 'min', 'max'])

print(result)
```

```python
df = pd.DataFrame({
    'school_code': ['s001', 's002', 's001', 's003', 's002'],
    'class': ['A', 'B', 'A', 'C', 'B'],
    'student': ['John', 'David', 'Alice', 'Peter', 'Sam'],
    'age': [15, 16, 15, 17, 16]
})

group = df.groupby(['school_code', 'class'])

# Display groups
for name, data in group:
    print("\nGroup:", name)
    print(data)
```
